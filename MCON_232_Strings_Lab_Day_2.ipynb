```python
from re import L
def read_books_no_csv(filename : str="books.csv"):
  fin = open(filename, "r")
  lines = fin.readlines()
  fin.close()

  books = []
  for line in lines[1:]:
    line = line.strip()

    if line == "":
      continue
    parts = line.split(",")

    title = parts[0].strip()
    author = parts[1].strip()
    year = int(parts[2].strip())
    country = parts[3].strip()
    genre = parts[4].strip()

    books.append((title, author, year, country, genre))

  return books

# print(read_books_no_csv())

csv_data = read_books_no_csv()
for book in csv_data:
  print(f"one book in our data: {book}")
```

```python
def read_books_csv_no_library(filename: str = "books.csv") -> list[dict[str, str]]:
  """
  Reads a simple CSV file (no quoted commas) without using the csv module.
  Returns a list of dictionaries: one dict per row.
  Assumes the first line is the header.
  """
  with open(filename, "r", encoding="utf-8") as file:
    lines = file. readlines ()

    # remove blank lines and strip newline characters
    lines = [line.strip() for line in lines if line.strip() ]

    header = lines [0].split(",")
    data_lines = lines[1:]

    books = []
    for line in data_lines:
      values = line.split(",")

      # build a dict for this row
      row_dict = {}
      for i in range(len(header)):
        row_dict [header[i] ] = values [i]

      books.append(row_dict)

  return books

books = read_books_csv_no_library("books.csv")
for book in books:
  print(f"{book}")
```

```python
def assignment(file_to_read, num_lines, file_to_write):
  lines = []
  with open(file_to_read, "r") as f:
    while num_lines > 0:
      for line in f:
        lines.append(line.strip())
        num_lines -= 1

  with open(file_to_write, "w") as g:
    for line in lines:
      g.write(line, "\n")


```

```python
def count_pale(file):
  with open(file, "r") as f:
    count = 0
    for line in f:
      cleaned = line.strip().replace(","," ").replace("."," ")
      words_as_lst = cleaned.split(" ")
      for word in words_as_lst:
        if word.startswith("pale") or word == "pallor":
          count += 1
  return count

print(count_pale("pg345_cleaned.txt"))

```
